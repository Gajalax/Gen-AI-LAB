```python
import torch
import transformers
import datasets

print("=" * 60)
print("SYSTEM CHECK")
print("=" * 60)

print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Datasets:", datasets.__version__)

print("\nCUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA version:", torch.version.cuda)
else:
    print("❌ GPU NOT AVAILABLE")
```

```python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding
)

print("=" * 60)
print("TRANSFORMERS IMPORT TEST")
print("=" * 60)

print("✅ AutoTokenizer")
print("✅ AutoModelForSequenceClassification")
print("✅ TrainingArguments")
print("✅ Trainer")
print("✅ DataCollatorWithPadding")
print("\n🎉 ALL IMPORTS WORKING!")
```

```python
from datasets import load_dataset

print("=" * 60)
print("LOADING IMDb DATASET")
print("=" * 60)

dataset = load_dataset("imdb")

train_ds = (
    dataset["train"]
    .shuffle(seed=42)
    .select(range(2000))
)

test_ds = (
    dataset["test"]
    .shuffle(seed=42)
    .select(range(500))
)

print("Training samples:", len(train_ds))
print("Testing samples:", len(test_ds))
```

```python
MODEL_NAME = "distilbert-base-uncased"

print("=" * 60)
print("LOADING DISTILBERT TOKENIZER")
print("=" * 60)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("✅ Tokenizer loaded successfully!")
```

```python
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=128
    )

print("Tokenizing training dataset...")

train_ds = train_ds.map(
    tokenize_function,
    batched=True
)

print("Tokenizing test dataset...")

test_ds = test_ds.map(
    tokenize_function,
    batched=True
)

print("✅ Tokenization completed!")
```

```python
data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

print("✅ Data collator created!")
```

```python
id2label = {
    0: "NEGATIVE",
    1: "POSITIVE"
}

label2id = {
    "NEGATIVE": 0,
    "POSITIVE": 1
}

print("=" * 60)
print("LOADING DISTILBERT")
print("=" * 60)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2,
    id2label=id2label,
    label2id=label2id
)

print("✅ DistilBERT model loaded!")
print("Number of labels:", model.config.num_labels)
```

```python
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(
        predictions,
        axis=1
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary",
        zero_division=0
    )

    accuracy = accuracy_score(
        labels,
        predictions
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("✅ Evaluation metrics configured!")
```

```python
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(
        predictions,
        axis=1
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary",
        zero_division=0
    )

    accuracy = accuracy_score(
        labels,
        predictions
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("✅ Evaluation metrics configured!")
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/distilbert_imdb_results",

    num_train_epochs=2,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    learning_rate=2e-5,
    weight_decay=0.01,

    eval_strategy="epoch",
    save_strategy="epoch",

    logging_strategy="steps",
    logging_steps=50,

    load_best_model_at_end=True,

    metric_for_best_model="accuracy",
    greater_is_better=True,

    fp16=True,

    report_to="none",

    seed=42
)

print("✅ Training arguments created successfully!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=test_ds,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

print("✅ Trainer created successfully!")
```

```python
import torch

print("=" * 60)
print("STARTING DISTILBERT FINE-TUNING")
print("=" * 60)

print("GPU:", torch.cuda.get_device_name(0))
print("Training samples:", len(train_ds))
print("Testing samples:", len(test_ds))
print("Epochs:", 2)

train_result = trainer.train()

print("\n" + "=" * 60)
print("FINE-TUNING COMPLETED")
print("=" * 60)
```

```python
print("=" * 60)
print("FINAL MODEL EVALUATION")
print("=" * 60)

metrics = trainer.evaluate()

print("\nEvaluation Results:")

for key, value in metrics.items():
    if isinstance(value, float):
        print(f"{key}: {value:.4f}")
    else:
        print(f"{key}: {value}")
```
